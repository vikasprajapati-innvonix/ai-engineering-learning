```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    temperature=0.7,
    max_tokens=200
)
```

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template("Explain {topic} to a {audience}")
prompt.invoke({"topic": "gravity", "audience": "beginner"})
```

```python
from langchain_core.prompts import ChatPromptTemplate

template = ChatPromptTemplate([
    ("system", "You are a helpful AI bot. Your name is Jarvis."),
    ("user","{user_input}")
])

template.invoke({"user_input": "Hello, there!"})
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

template = ChatPromptTemplate([
    ("system", "You are a helpful AI bot. Your name is Jarvis."),
    MessagesPlaceholder("chat_history"),
    ("human","{question}")
])

template.invoke({
    "chat_history": [("human", "What is a list?"), ("ai", "An ordered collection.")],
    "question": "Hello, there!"
})
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt = ChatPromptTemplate.from_template("Tell me a fun fact about {topic}")
parser = StrOutputParser()

formatted_message = prompt.format_messages(topic="pandas")
# llm.invoke(formatted_message


chain = prompt | llm | parser
chain.invoke({"topic": "pandas"})
```
