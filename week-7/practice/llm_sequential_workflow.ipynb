```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
from langchain_groq import ChatGroq
```

```python
llm_model = ChatGroq(model="openai/gpt-oss-120b", max_tokens = 1000)
```

```python
class BlogState(TypedDict):
    topic: str
    outline: str
    blog: str
```

```python
def generate_blog_outline(state: BlogState) -> BlogState:
    topic = state["topic"]
    
    prompt = f"Write a blog outline about {topic}"
    outline = llm_model.invoke(prompt)
    state["outline"] = outline.content
    return state
```

```python
def generate_blog(state: BlogState) -> BlogState:
    topic = state["topic"]
    outline = state["outline"]
    
    prompt = f"Write a blog about {topic} based on the outline {outline}"
    blog = llm_model.invoke(prompt)
    state["blog"] = blog.content
    return state
```

```python
graph = StateGraph(BlogState)

graph.add_node("generate_blog_outline", generate_blog_outline)
graph.add_node("generate_blog", generate_blog)

graph.add_edge(START, "generate_blog_outline")
graph.add_edge("generate_blog_outline", "generate_blog")
graph.add_edge("generate_blog", END)

workflow = graph.compile()
```

```python
topic = "AI impact on the society"

initial_state = {
    "topic": topic,
    "outline": "",
    "blog": ""
}

state = workflow.invoke(initial_state)
```

```python
from IPython.display import display, Markdown

display(Markdown("Topic"))
display(Markdown(state["topic"]))
display(Markdown("Outline"))
display(Markdown(state["outline"]))
display(Markdown("Blog"))
display(Markdown(state["blog"]))
```
