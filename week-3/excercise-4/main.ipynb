```python
from langchain_ollama import ChatOllama
import time

llm = ChatOllama(model="llama3")

start = time.time()
for chunk in llm.stream("Write an article about how does global warming impacting human lifes."):
    print(chunk.text, end="", flush=True)
end = time.time()

print(f'Total time take {end - start}')
```
