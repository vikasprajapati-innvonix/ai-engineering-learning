```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_mistralai import MistralAIEmbeddings

embedding = MistralAIEmbeddings(
    model="mistral-embed"
)
```

```python
documents = [
    "New Delhi is capital of India",
    "Washington, D.C. is capital of United State of America",
    "Beijing is capital of China",
    "London is capital of United Kingdom",
    "Tokyo is capital of Japan"
]    
```

```python
from langchain_community.vectorstores import FAISS

db = FAISS.from_texts(documents,embedding=embedding)
```

```python
print(db)
```

```python
results = db.similarity_search("what is capital of japan")
print(results[0].page_content)
```
