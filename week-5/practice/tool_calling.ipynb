```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool

@tool("multiply",description="Multiply two numbers")
def multiply(a:int,b:int) -> int: 
    return a * b
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(model='openai/gpt-oss-20b',max_tokens=200,temperature=0.1)

llm_with_tools = llm.bind_tools([multiply])
```

```python
llm_with_tools.invoke("how are you?")
```

```python
llm_with_tools.invoke('What 30 * 2')
```

```python
from langchain_core.messages import HumanMessage
query = HumanMessage('50 * 50')
result = llm_with_tools.invoke([query])

print(result.tool_calls)
multiply_tool_calls = result.tool_calls[0]

multiply.invoke(multiply_tool_calls)
```
